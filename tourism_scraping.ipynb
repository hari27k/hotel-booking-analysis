## Web Scraping Using Beautiful Soup

### Import Libraries

```python
import requests
from bs4 import BeautifulSoup
```

```python
import pandas as pd
```

### Initializing the Soup

```python
url="https://en.wikipedia.org/wiki/World_Tourism_rankings"
headers = {
    "User-Agent": "Mozilla/5.0"
}
```

```python
response = requests.get(url,headers=headers)
```

```python
response.status_code
```

```python
Soup = BeautifulSoup(response.text,"html.parser")
```

### a. find() or find_all()

```python
table = Soup.find("table")
```

```python
rows = table.find_all("tr")
```

```python
data =[]
```

### b. Loops

```python
for row in rows[1:11]:   
    columns = row.find_all(["th", "td"])

    if len(columns) >= 2:
        country = columns[0].get_text(strip=True)
        tourist_arrivals = columns[1].get_text(strip=True)

        data.append({
            "Country": country,
            "Tourism Rank or Visitor Information": tourist_arrivals
        })
```

### c. Pandas Dataframe

```python
df_tourism = pd.DataFrame(data)
```

```python
print(df_tourism)
```

```python
df_tourism.to_csv("tourism_data.csv", index=False)

print("\nFile saved successfully as tourism_data.csv")
```
